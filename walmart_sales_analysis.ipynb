### Simple Population Example

Imagine we have data on the **elderly population (aged 65+)** in a region over several years. We want to look at how the elderly population is changing because this may help us understand future healthcare demand.

| Year | Population aged 65+ |
|------|----------------------:|

```python
years = np.arange(2021, 2026)

elderly_population = np.array([20000, 21000, 22500, 24000, 26000])

print(years)
print(elderly_population)
```

#### Visualizing the Population Data

A line plot and scatter plot are used to visualize the population values over time. The x-axis represents the year, while the y-axis represents the population aged 65+.

```python
plt.plot(years, elderly_population)
plt.scatter(years, elderly_population)

plt.xlabel('Year')
plt.ylabel('Population aged 65+')
plt.title('Elderly Population Forecasting')
plt.xticks(years)

plt.show()
```

#### Observation

The graph shows an increasing trend in the imaginary elderly population. The population increases from 20,000 in 2021 to 26,000 in 2025. This simple example confirms that the Python environment, NumPy, and Matplotlib are working correctly.

### 2.2 Python Programming Basics

The following examples demonstrate four basic Python programming concepts: a variable, a function, a conditional statement, and a loop. The examples are connected to the elderly population scenario used in the previous section.

#### Variable

A variable is used to store data that can be used later in the program. Here, `elderly_population` stores the imaginary population values for people aged 65+.

```python
elderly_population = np.array([20000, 21000, 22500, 24000, 26000])

print(elderly_population)
```

#### Function

A function is a reusable block of code that performs a specific task. The `population_growth()` function calculates the increase between an old population value and a new population value.

```python
def population_growth(old_population, new_population):
    return new_population - old_population

print(population_growth(24000, 26000))
```

#### Conditional Statement

A conditional statement checks a condition and decides which code should run. Here, the program checks whether the latest elderly population is above 25,000.

```python
if elderly_population[-1] > 25000:
    print('Elderly population is above 25,000')
else:
    print('Elderly population is below 25,000')
```

#### Loop

A loop repeats the same action for each item in a sequence. Here, the `for` loop goes through the elderly population data and prints each population value.

```python
for population in elderly_population:
    print(population)
```

## Part 3: Dataset Exploration Using Walmart Sales Data

The Walmart dataset contains weekly sales information for different stores. In this section, one store is randomly selected using my student ID as the random seed. The selected store is then explored using summary statistics and a sales trend graph.

### 3.1 Load and Subset the Data

The Walmart dataset is loaded using Pandas. My student ID (291013) is used as the random seed to randomly select one store. The dataset is then filtered to create a subset containing only the selected store.

```python
import pandas as pd
import random

df = pd.read_csv("Walmart.csv")

random.seed(291013)

my_store = random.choice(df['Store'].unique())

df_subset = df[df['Store'] == my_store].reset_index(drop=True)
```

```python
print('My selected store:', my_store)
```

### 3.2 Explore the Structure

The first five rows of the selected store are displayed to get an initial view of the data. The `.info()` method is used to examine the structure and data types, while `.describe()` provides summary statistics for the numerical variables.

```python
df_subset.head()
```

```python
df_subset.info()
```

#### Data Structure Observation

The selected store contains 143 observations and 8 columns. There are no missing values, as all columns have 143 non-null entries. The dataset contains integer, float, and object data types. The `Date` column is currently stored as an object and will be converted to a date format before plotting.

```python
df_subset.describe()
```

#### Summary Statistics Observation

Store 42 contains 143 weekly sales observations. The average weekly sales are approximately 556,404, with sales ranging from about 428,954 to 674,919. The median weekly sales are approximately 556,046, which is close to the mean. The average temperature is 72.24, while the average fuel price is 3.58. The Holiday Flag has a mean of 0.07, indicating that most observations are non-holiday weeks.

### 3.3 Plot the Trend

To observe the sales pattern of Store 42 over time, Weekly Sales are plotted against Date. The Date column is first converted from text into a date format so that the values are displayed correctly on the time axis.

```python
# 3.3 Plot Weekly Sales over Date

df_subset['Date'] = pd.to_datetime(df_subset['Date'], dayfirst=True)

plt.plot(df_subset['Date'], df_subset['Weekly_Sales'], label='Weekly Sales')

plt.xlabel('Date')
plt.ylabel('Weekly Sales')
plt.title('Weekly Sales Trend for Store ' + str(my_store))
plt.legend()

plt.xticks(rotation=45)

plt.show()
```

#### Sales Trend Observation

The graph shows that weekly sales for Store 42 fluctuate over time. There are several noticeable increases and decreases in sales, but the overall pattern does not show a continuous upward or downward trend. Some larger changes may represent unusual sales weeks, which are investigated further in the next section.

### 3.4 Identify Spikes and Patterns

To investigate unusual sales, the highest and lowest weekly sales are identified. The `Holiday_Flag` is also checked to see whether these sales occurred during holiday weeks.

```python
highest_sales = df_subset.loc[df_subset['Weekly_Sales'].idxmax()]
lowest_sales = df_subset.loc[df_subset['Weekly_Sales'].idxmin()]

print('Highest sales week:')
print(highest_sales[['Date', 'Weekly_Sales', 'Holiday_Flag']])

print('\nLowest sales week:')
print(lowest_sales[['Date', 'Weekly_Sales', 'Holiday_Flag']])
```

The code finds the weeks with the maximum and minimum Weekly Sales. It also displays the date and Holiday Flag for these weeks. A Holiday Flag of 1 represents a holiday week, while 0 represents a non-holiday week.

#### Observation

Store 42 shows frequent changes in weekly sales, with several noticeable peaks and drops. The highest weekly sales were approximately 674,919, while the lowest were approximately 428,954. Both occurred during holiday weeks, as shown by a Holiday Flag of 1. This suggests that holidays may be related to unusual sales patterns. However, holiday weeks do not always lead to higher sales, since both the highest and lowest sales occurred during holidays. Overall, sales fluctuate over time without a clear continuous upward or downward trend.
